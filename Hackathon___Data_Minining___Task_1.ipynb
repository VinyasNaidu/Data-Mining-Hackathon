```python
import pandas as pd
import re


# Helper function to clean subcategory (level 1) header text.
def clean_subheader(text):
    if pd.isna(text):
        return ""
    # Remove newline characters, slashes, and collapse extra whitespace.
    cleaned = re.sub(r'[\n/\\]', ' ', str(text))
    cleaned = re.sub(r'\s+', ' ', cleaned)
    return cleaned.strip()

# Step 2: Process the MultiIndex columns.
#         - Use the last valid top-level header when a cell shows as "Unnamed".
#         - Clean the subcategory header.
#         - If the subcategory header is missing/unnamed, use the top-level header.
new_columns = []
last_top = None  # To store the last valid big category header

for top, sub in df.columns:
    # Fill down missing top-level header (if "Unnamed" or NaN)
    if pd.isna(top) or "Unnamed" in str(top):
        top = last_top
    else:
        last_top = top

    # Clean the subcategory text.
    sub_cleaned = clean_subheader(sub)

    # If the subcategory is empty or "Unnamed", replace it with the top-level header.
    if not sub_cleaned or "Unnamed" in str(sub):
        sub_cleaned = str(top).strip()

    new_columns.append((str(top).strip(), sub_cleaned))

# Reassign the new MultiIndex columns to the DataFrame.
df.columns = pd.MultiIndex.from_tuples(new_columns)

```

```python
# --- Step 1: Filter for Universities ---
uni_df = df[df[('Agency Type', 'Agency Type')] == 'Universities and Colleges'].copy()

# --- Step 2: Create a grouping column based on the Agency Name ---
# Adjust the check if the naming in your data is different.
uni_df['University_Group'] = uni_df[('Agency Name', 'Agency Name')].astype(str).apply(
    lambda x: 'Michigan State University' if "Michigan State University" in x else 'Other Universities'
)

# --- Step 3: Define the aggregate columns for each crime category ---
# Crimes Against Persons aggregates (including both "Sex Offenses" aggregates)
cap_persons_cols = [
    ('Crimes Against Persons', 'Assault Offenses'),
    ('Crimes Against Persons', 'Homicide Offenses'),
    ('Crimes Against Persons', 'Kidnapping Abduction'),
    ('Crimes Against Persons', 'Sex Offenses'),
    ('Crimes Against Persons', 'Sex Offenses, Nonforcible')
]

# Crimes Against Property aggregates
cap_property_cols = [
    ('Crimes Against Property', 'Arson'),
    ('Crimes Against Property', 'Bribery'),
    ('Crimes Against Property', 'Burglary Breaking & Entering'),
    ('Crimes Against Property', 'Counter- feiting Forgery'),
    ('Crimes Against Property', 'Destruction Damage Vandalism'),
    ('Crimes Against Property', 'Embezzle- ment'),
    ('Crimes Against Property', 'Extortion Blackmail'),
    ('Crimes Against Property', 'Fraud Offenses'),
    ('Crimes Against Property', 'Larceny- Theft Offenses'),
    ('Crimes Against Property', 'Motor Vehicle Theft'),
    ('Crimes Against Property', 'Robbery'),
    ('Crimes Against Property', 'Stolen Property Offenses')
]

# Crimes Against Society aggregates
cap_society_cols = [
    ('Crimes Against Society', 'Drug Narcotic Offenses'),
    ('Crimes Against Society', 'Gambling Offenses'),
    ('Crimes Against Society', 'Porno- graphy Obscene Material'),
    ('Crimes Against Society', 'Prost- itution Offenses'),
    ('Crimes Against Society', 'Weapon Law Violations')
]

# --- Step 4: Define a helper function to compute totals over given columns ---
def compute_category_totals(df, cols):
    total = 0
    for col in cols:
        # Remove commas and convert to numeric (errors become NaN, which sum() skips)
        numeric_col = pd.to_numeric(
            df[col].astype(str).str.replace(',', '', regex=True),
            errors='coerce'
        )
        total += numeric_col.sum(skipna=True)
    return total

# --- Step 5: Group by our new University_Group and compute aggregate totals ---
grouped = uni_df.groupby('University_Group')

results = {}
for group_name, group_df in grouped:
    total_persons = compute_category_totals(group_df, cap_persons_cols)
    total_property = compute_category_totals(group_df, cap_property_cols)
    total_society = compute_category_totals(group_df, cap_society_cols)
    total_offenses = total_persons + total_property + total_society
    results[group_name] = {
        'Crimes Against Persons': total_persons,
        'Crimes Against Property': total_property,
        'Crimes Against Society': total_society,
        'Total Offenses': total_offenses
    }

# Create a pivot DataFrame from the results
pivot_df = pd.DataFrame(results).T

print("Offenses Comparison between Michigan State University and Other Universities:")
pivot_df
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler, PolynomialFeatures

# ---------------------------
# Step 2: Baseline Linear Regression
# ---------------------------
# Ensure no NaNs remain in the scaled training and testing sets.
X_train = np.nan_to_num(X_train, nan=0.0)
X_test = np.nan_to_num(X_test, nan=0.0)

lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
mse_lr = mean_squared_error(y_test, y_pred_lr)
r2_lr = r2_score(y_test, y_pred_lr)
print("Linear Regression - MSE:", mse_lr, "R^2:", r2_lr)



```
